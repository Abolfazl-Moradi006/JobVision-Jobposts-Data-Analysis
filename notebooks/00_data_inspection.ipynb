### Build the Date Tools

```python
import jdatetime

col = "Jobpost_ActivationTime_YEAR_MONTH"

# Gregorian date, year, month
jobposts_clean["date"] = pd.to_datetime(jobposts_clean[col], format="%Y-%m")
jobposts_clean["year"] = jobposts_clean["date"].dt.year
jobposts_clean["month"] = jobposts_clean["date"].dt.month 

# We only know the month, so we use the 15th as an approximate day.
months = jobposts_clean["date"].drop_duplicates()
jalali = {d: jdatetime.date.fromgregorian(date=d.replace(day=15).date()) for d in months}
jobposts_clean["jalali_year"] = jobposts_clean["date"].map(lambda d: jalali[d].year)
jobposts_clean["jalali_month"] = jobposts_clean["date"].map(lambda d: jalali[d].month)

# Jalali season
season_map = {**dict.fromkeys([1, 2, 3], "Spring"),
              **dict.fromkeys([4, 5, 6], "Summer"),
              **dict.fromkeys([7, 8, 9], "Autumn"),
              **dict.fromkeys([10, 11, 12], "Winter")}
jobposts_clean["jalali_season"] = jobposts_clean["jalali_month"].map(season_map)

display((jobposts_clean.groupby([col, "jalali_year", "jalali_month", "jalali_season"])
   .size().rename("ads").reset_index()))
```

#### Date Tools

Created `date`, `year`, `month`, `jalali_year`, `jalali_month` and `jalali_season` from `Jobpost_ActivationTime_YEAR_MONTH`. The period column will be added later when needed.

**Findings:**
- The data covers 25 months, from 2022-03 to 2024-03, with no gaps.
- In Persian years this is **1401 and 1402** (plus the last month of 1400). It does **not** reach 1403, because 2024-03 still falls in Esfand 1402. So the brief's "1402-1403" does not match the data, and we trust the data.
- Ads per month are higher in 2023 than in 2022 (about 16-24k vs 10-16k).
- Three months have unusually low counts: 2022-03 (901), 2023-03 (5,531) and 2024-03 (3,854). The first and last are probably partial months at the edges of the data. Be careful with them in the seasonality analysis.

**Note:** Jalali months are **approximate**. We only know the Gregorian month, so the 15th was used as the day, and each Gregorian month is split across two Persian months.

### Dataset Structure

Each row is one job ad. The columns fall into five groups: job details (title, province, work type, salary, experience, working hours, month), job categories, skills and qualifications (three JSON-text columns), benefits and policies, and company details (size, age, ownership, location, industry). Full column definitions are in `Data_Dictionary.md`.

### Main Limitations

- **Missing salary:** about 67% of ads have no salary, so all salary results describe only the ads that show it. Whether the missing ones are random is studied in the next issue.
- **No company ID and no real post ID:** we cannot group ads by employer, and `RowNumber` is only a counter.
- **Short time span:** 25 months is too short for long-term trends. The first and last months (2022-03 and 2024-03) have very few ads and are probably partial.
- **Approximate Jalali dates:** only the month is known, so Persian months are approximate.
- **Offered values, not outcomes:** salary, age and gender columns show what employers ask for or offer, not who was hired or what they were paid.
- **Unparsed text:** the JSON columns and the working-hours text are not parsed yet. This is done in the feature engineering issue.
- **Duplicates:** only fully identical rows were checked (none found). Near-duplicates, such as repeated ads from the same company, were not checked.

### Conclusion

The base file `data/interim/jobvision_base.parquet` is ready for the next issues. The data is well structured and consistent in its categories, and the main weakness is the large share of missing salaries.
